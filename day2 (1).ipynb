```python
#even numbers from 1 to 50
g=np.arange(1,50,2)
g
```

```python
#multiple of 5
h=np.arange(0,100,5)
print(h)
#maximum and minimum of array
print(max(h))
print(min(h))
```

```python
#largest element of index position
i=np.array([20,56,78,43,22])
np.argmax(i)
```

```python
#ascending order
j=np.array([20,56,78,43,22])
np.sort(j)
```

```python
#reverse without loops
k=np.array([2,5,6,7,8])
k[::-1]
```

```python
#element wise addition of two arrays
l=np.array([35,67,89,45])
m=np.array([43,56,7,8])
np.dot(l,m)
```

```python
#sub,mul,div on arrays
n=np.array([5,6,7,8])
o=np.array([2,3,3,4])
p1=n-o
p2=n*o
p3=n/o
print(p1,p2,p3)
```

```python
#mean,median,SD
p=np.array([5,6,7,8])
print(np.mean(p))
print(np.median(p))
print(np.std(p))
```

```python
#sum of rows and columns
q=np.array([[3,4,5],[6,7,8],[9,3,5]])
print(np.sum(q,axis=0)) #column
print(np.sum(q,axis=1))#row
```

```python
#mul of two matrices
r=np.array([[3,4],[1,2]])
s=np.array([[2,1],[3,2]])
print(np.dot(r,s))
```
