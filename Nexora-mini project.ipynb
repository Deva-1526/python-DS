```python
import pandas as pd
```

```python
d1=pd.read_csv('lat_UE10_SCS15_CAP1_MCSLEP_BW20_PKT1200_SINR15p0_BR10_TNmid_CNcloud_R3_SEED42 (1).csv')
d2=pd.read_csv('lat_UE10_SCS15_CAP1_MCSLEP_BW20_PKT1200_SINR5p0_BR10_TNmid_CNcloud_R3_SEED42.csv')
d3=pd.read_csv('lat_UE10_SCS15_CAP2_MCSLEP_BW20_PKT1200_SINR5p0_BR10_TNmid_CNcloud_R3_SEED42.csv')
df=pd.concat([d1,d2,d3],ignore_index=True)
```

```python
df.drop(['ue_id', 'packet_idx','success', 'dropped', 'ue_count', 'scs_khz','mcs_mode', 'bandwidth_hz', 'packet_size_bits',
       'duration_sec', 'max_retries', 'scheduler_type', 'tn_mode', 'core_mode',
       'lambda_packets_per_s', 'seed', 'bitrate_bps','avg_latency_ms_ue', 'latency_90th_ms_ue', 'latency_std_ms_ue',
       'packet_loss_rate_ue', 'eff_throughput_mbps_ue'],axis="columns")
```

```python
from sklearn.model_selection import train_test_split
x=df[["arrival_time_ms","ue_capabilities","sinr_db","scenario_name"]]
y=df["latency_e2e_ms"]
x=pd.get_dummies(x,columns=["scenario_name"])
print(x.head())
print(x.shape)
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train,y_test = train_test_split(x,y, test_size=0.2, random_state=42)
print("Traning data:",x_train.shape)
print("Testing data:",x_test.shape)
```

```python
from sklearn.tree import DecisionTreeRegressor
model=DecisionTreeRegressor(max_depth=10,random_state=42)
model.fit(x_train,y_train)
y_pred = model.predict(x_test)
```

```python
print(y_pred[:10])
```

```python
from sklearn.metrics import mean_squared_error
from sklearn.metrics import mean_absolute_error
from sklearn.metrics import r2_score
import numpy as np
mae=mean_absolute_error(y_test,y_pred)
rmse=np.sqrt(mean_squared_error(y_test,y_pred))
r2=r2_score(y_test,y_pred)
print("MAE:",mae)
print("RMSE:",rmse)
print("R2 Score:",r2)
```

```python
from sklearn.tree import DecisionTreeRegressor
from sklearn.metrics import r2_score
for depth in [5,10,15,20]:
    model = DecisionTreeRegressor(max_depth=depth,random_state=42)
    model.fit(x_train,y_train)
    y_pred=model.predict(x_test)
    r2=r2_score(y_test,y_pred)
    print("Depth:",depth,"r2_score:",r2)
```

```python
from sklearn.tree import DecisionTreeRegressor
from sklearn.metrics import mean_squared_error,mean_absolute_error,r2_score
import numpy as np
final_model=DecisionTreeRegressor(max_depth=10,random_state=42)
final_model.fit(x_train,y_train)
final_pred=final_model.predict(x_test)
mae=mean_absolute_error(y_test,final_pred)
rmse=np.sqrt(mean_squared_error(y_test,final_pred))
r2=r2_score(y_test,final_pred)
print("Final Model Performance")
print("MAE:",mae)
print("RMSE:",rmse)
print("R2 Score:",r2)
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(8,5))
plt.scatter(y_test,final_pred,alpha=0.3)
plt.xlabel("Actual latency(ms)")
plt.ylabel("Predicted latency(ms)")
plt.title("Actual vs Predicted IoT Network latency")
plt.show()
```
