```python
import matplotlib.pyplot as plt
```

```python
#line chart
plt.plot([4.5,6,7,4,6],marker='d',linestyle='dashdot',ms=10,mfc='pink',mec='violet',linewidth=3)
plt.show()
```

```python
#bar chart
gadgets=['mobile','tab','laptop','smart watch']
price=[20000,15000,60000,2000]
plt.bar(gadgets,price,color=['pink','blue','yellow','green'])
plt.show()
```

```python
#pie chart
gadgets=['mobile','tab','laptop','smart watch']
price=[20000,15000,60000,2000]
plt.pie(price,labels=gadgets,colors=['pink','blue','yellow','green'],autopct='%1.1f%%')
plt.show()
```

```python
#histogram
marks=[45,67,78,93,55,99]
plt.hist(marks,color='purple')
plt.show()
```

```python
import pandas as pd
d=pd.read_csv("archive (5).zip")
d
```

```python
plt.hist(d['Age'])
plt.show()
```

```python
plt.bar(d['Age'],d['Survived'])
plt.xlabel("Age")
plt.ylabel("Survived")
plt.title("Age VS Survived")
plt.show()
```

```python
#scatter plot
plt.scatter(d['Age'],d['Fare'])
plt.show()
```

```python
d['Age'].isnull().sum()
```

```python
d['Age'].fillna(d['Age'].mean(),inplace=True)
print(d['Age'].isnull().sum())
```

```python
d.boxplot(column='Age', by='Survived')
plt.title('Box Plot of Age by Survived')
plt.ylabel('Age')
plt.show()
```

```python
#area chart
plt.fill_between(d['Age'],d['Survived'])
plt.show()
```

```python
#regex
import re
```

```python
text="my number is 1234567890"
n=re.findall(r"\d",text)
print(n)
```

```python
d="Data science is my fav subject"
e=re.search("Data science",d)
print(e)
print(e.group())
```

```python
d="i like Data science"
match=re.match("Data science",d)

print(match)
```

```python
#search
txt="The pen is yours"
x=re.split("\s",txt)
print(x)
```

```python
#sub
txt="The pen is yours"
x=re.sub("\s","9",txt)
print(x)
```
